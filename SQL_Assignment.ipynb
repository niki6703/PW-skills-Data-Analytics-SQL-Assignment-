# **What Are Constraints in SQL?**

Constraints are rules applied to columns in a table to restrict the type of data that can be stored in them.

 Purpose: To enforce data integrity, prevent invalid data entry, and ensure database accuracy and reliability.

# How Constraints Help Maintain Data Integrity

Constraints:

Prevent duplicate or invalid values.

Ensure relationships between tables are valid.

Protect critical fields from being empty.

Enforce business rules, like minimum age or salary.

# Primary Keys and Foreign Keys in the Maven Movies Database
In the context of a Maven Movies database, primary keys (PK) and foreign keys (FK) play crucial roles in defining data relationships and ensuring data integrity.

**Primary Keys (PK):**
A primary key is a unique identifier for each record in a table. It ensures that each record can be uniquely identified by its primary key column. Primary keys cannot contain NULL values, and each record must have a unique value for this key.

**In the Maven Movies database:**

**Movies Table:** MovieID is the primary key, uniquely identifying each movie.

**Directors Table:** DirectorID is the primary key, uniquely identifying each director.

**Actors Table:** ActorID is the primary key, uniquely identifying each actor.

**Foreign Keys (FK):**
A foreign key is a column (or set of columns) that creates a relationship between two tables. It references the primary key of another table, ensuring referential integrity between the tables. A foreign key allows a record in one table to be linked to a record in another table. Foreign keys can contain NULL values, indicating that a relationship is not mandatory for all records.

**In the Maven Movies database:**

**Movies Table:** DirectorID is a foreign key referencing the DirectorID in the Directors table, linking each movie to its director.

**MovieActors Table:** Both MovieID and ActorID are foreign keys. MovieID references the MovieID in the Movies table, and ActorID references the ActorID in the Actors table, establishing a many-to-many relationship between movies and actors.

# Differences Between Primary Keys and Foreign Keys:
Uniqueness: A primary key ensures that each value is unique within its table, while a foreign key can have repeated values.

Nullability: A primary key cannot have NULL values, whereas a foreign key can have NULL values.

Purpose: The primary key uniquely identifies records within its own table. The foreign key is used to establish and enforce relationships between tables.

Location: The primary key is located within the same table, while the foreign key references a primary key in another table.

By defining primary and foreign keys, the database maintains data integrity and establishes meaningful relationships between different entities like movies, directors, and actors.

Que 8. Calculate the running total of rentals per category, ordered by rental count.

**1. First Normal Form (1NF):**

a. Identify a table in the Sakila database that violates 1NF. Explain how you

# Ans:> First Normal Form (1NF):
**Definition:**

A table is in First Normal Form (1NF) if:

All columns contain only atomic (indivisible) values.

There are no repeating groups or arrays in a single column.

Each record is unique.

Example: A denormalized "rental_summary" table

| customer_id | customer_name | rented_films                       |
|-------------|----------------|------------------------------------|
| 1           | John Doe       | "Shrek, Batman Begins, Toy Story" |
| 2           | Jane Smith     | "Inception, Matrix"               |

Here, the rented_films column stores multiple values (i.e., a list of films) in a single field, which violates 1NF because the values are not atomic.

**Violation of 1NF:**

The rented_films column contains comma-separated lists, which are non-atomic.

You cannot efficiently query or update individual films in this format.

# **How to normalize this table to achieve 1NF:**

We create a new table where each film rented is stored in a separate row:

**Normalized Rental Table (In 1NF):**

| customer_id | customer_name | rented_film       |
|-------------|----------------|-------------------|
| 1           | John Doe       | Shrek             |
| 1           | John Doe       | Batman Begins     |
| 1           | John Doe       | Toy Story         |
| 2           | Jane Smith     | Inception         |
| 2           | Jane Smith     | Matrix            |

Now:

Each column contains atomic values.

There are no repeating groups.

The table satisfies 1NF.




Que 2. Second Normal Form (2NF):

a. Choose a table in Sakila and describe how you would determine whether it is in 2NF.
If it violates 2NF, explain the steps to normalize it.

# **Ans:>**
 **What is Second Normal Form (2NF)?**

A table is in 2NF if:

It is already in 1NF (First Normal Form), and

There are no partial dependencies — i.e., no non-prime attribute depends only on part of a composite primary key.

**Suppose we have this badly designed table:**

film_rental_info (violates 2NF):





```python
CREATE TABLE film_rental_info (
    film_id INT,
    store_id INT,
    film_title VARCHAR(100),
    category_name VARCHAR(50),
    rental_rate DECIMAL(4,2),
    PRIMARY KEY (film_id, store_id)
);
```

Here:

film_id and store_id together form the composite primary key.

But:

film_title, category_name, and rental_rate depend only on film_id, not on store_id.

These are partial dependencies, so this table violates 2NF.

#  How to Normalize to 2NF:
Step 1: Split the table into two tables:

Table 1: film_details (stores film info)


```python
CREATE TABLE film_details (
    film_id INT PRIMARY KEY,
    film_title VARCHAR(100),
    category_name VARCHAR(50),
    rental_rate DECIMAL(4,2)
);
```

Table 2: film_store (only holds the relationship between film and store)

```python
CREATE TABLE film_store (
    film_id INT,
    store_id INT,
    PRIMARY KEY (film_id, store_id),
    FOREIGN KEY (film_id) REFERENCES film_details(film_id)
);
```

Now:

film_details contains only attributes fully dependent on film_id.

film_store represents the many-to-many relationship.

No partial dependencies → Now in 2NF.

Que 3. Third Normal Form (3NF):
a. Identify a table in Sakila that violates 3NF. Describe the transitive dependencies
present and outline the steps to normalize the table to 3NF.


# Ans:>
**What is Third Normal Form (3NF)?**

A table is in 3NF if:

It is already in 2NF, and

It does not contain transitive dependencies — i.e., non-prime attributes should not depend on other non-prime attributes.

Let’s suppose we have a poorly designed version of the customer table like this:

 **Table: customer_info (violates 3NF)**



```python
CREATE TABLE customer_info (
    customer_id INT PRIMARY KEY,
    first_name VARCHAR(50),
    last_name VARCHAR(50),
    email VARCHAR(100),
    address VARCHAR(100),
    city VARCHAR(50),
    country VARCHAR(50)
);
```

**🔴 Transitive Dependencies:**

customer_id → address

address → city

city → country

Here:

country depends on city, and city depends on address, which depends on customer_id.

So, country and city are transitively dependent on the primary key (customer_id).

Thus, this table violates 3NF.

#  Steps to Normalize into 3NF:
Step 1: Break the transitive dependencies into separate tables.

Table 1: customer

```python
CREATE TABLE customer (
    customer_id INT PRIMARY KEY,
    first_name VARCHAR(50),
    last_name VARCHAR(50),
    email VARCHAR(100),
    address_id INT
);
```

Table 2: address

```python
CREATE TABLE address (
    address_id INT PRIMARY KEY,
    address VARCHAR(100),
    city_id INT
);
```

Table 3: city

```python
CREATE TABLE city (
    city_id INT PRIMARY KEY,
    city_name VARCHAR(50),
    country_id INT
);
```

Table 4: country

```python
CREATE TABLE country (
    country_id INT PRIMARY KEY,
    country_name VARCHAR(50)
);
```

Now:

All non-prime attributes directly depend on their table's primary key.

No transitive dependencies.

This structure is in Third Normal Form (3NF).

# 5. CTE Basics:
Que a. Write a query using a CTE to retrieve the distinct list of actor names and the number of films they
have acted in from the actor and film_actor tables.

```python
-- Using Common Table Expression (CTE)
WITH actor_film_count AS (
    SELECT
        a.actor_id,
        CONCAT(a.first_name, ' ', a.last_name) AS actor_name,
        COUNT(fa.film_id) AS film_count
    FROM actor a
    JOIN film_actor fa ON a.actor_id = fa.actor_id
    GROUP BY a.actor_id, a.first_name, a.last_name
)

SELECT * FROM actor_film_count
ORDER BY film_count DESC;
```

**Explanation:**

We use a CTE named actor_film_count to:

Join actor and film_actor tables.

Count how many films each actor has acted in.

Concatenate first_name and last_name for a complete name.

Finally, we select from the CTE and order results by film count (optional).

# 7. CTE for Aggregation:
a. Write a query using a CTE to find the total revenue generated by each customer (sum of payments)

```python
-- Common Table Expression to calculate total revenue per customer
WITH customer_revenue AS (
    SELECT
        c.customer_id,
        CONCAT(c.first_name, ' ', c.last_name) AS customer_name,
        SUM(p.amount) AS total_revenue
    FROM customer c
    JOIN payment p ON c.customer_id = p.customer_id
    GROUP BY c.customer_id, c.first_name, c.last_name
)

SELECT *
FROM customer_revenue
ORDER BY total_revenue DESC;
```

# 9. CTE and Filtering:
a. Create a CTE to list customers who have made more than two rentals, and then join this CTE with the
customer table to retrieve additional customer details.

```python
-- CTE to find customers with more than 2 rentals
WITH frequent_customers AS (
    SELECT
        customer_id,
        COUNT(rental_id) AS rental_count
    FROM rental
    GROUP BY customer_id
    HAVING COUNT(rental_id) > 2
)

-- Join with customer table to get additional details
SELECT
    fc.customer_id,
    CONCAT(c.first_name, ' ', c.last_name) AS customer_name,
    c.email,
    fc.rental_count
FROM frequent_customers fc
JOIN customer c ON fc.customer_id = c.customer_id
ORDER BY fc.rental_count DESC;
```

Explanation:

The CTE frequent_customers:

Groups the rental table by customer_id.

Filters customers with more than 2 rentals using the HAVING clause.

The final query joins this CTE with the customer table to get:

Full name,

Email,

Rental count.

# 10. CTE for Date Calculations:
a. Write a query using a CTE to find the total number of rentals made each month, considering the
rental_date from the rental table

```python
-- CTE to calculate total rentals per month
WITH monthly_rentals AS (
    SELECT
        DATE_FORMAT(rental_date, '%Y-%m') AS rental_month,
        COUNT(rental_id) AS total_rentals
    FROM rental
    GROUP BY DATE_FORMAT(rental_date, '%Y-%m')
)

SELECT *
FROM monthly_rentals
ORDER BY rental_month;
```

Explanation:

The CTE monthly_rentals:

Extracts the month and year from the rental_date using DATE_FORMAT('%Y-%m').

Counts how many rentals occurred in each month.

# 11. CTE and Self-Join:
a. Create a CTE to generate a report showing pairs of actors who have appeared in the same film
together, using the film_actor table.

```python
-- CTE to list all pairs of actors who appeared in the same film
WITH actor_pairs AS (
    SELECT
        fa1.film_id,
        fa1.actor_id AS actor1_id,
        fa2.actor_id AS actor2_id
    FROM film_actor fa1
    JOIN film_actor fa2
        ON fa1.film_id = fa2.film_id
        AND fa1.actor_id < fa2.actor_id
)

-- Final query joining actor names for better readability
SELECT
    ap.film_id,
    CONCAT(a1.first_name, ' ', a1.last_name) AS actor1_name,
    CONCAT(a2.first_name, ' ', a2.last_name) AS actor2_name
FROM actor_pairs ap
JOIN actor a1 ON ap.actor1_id = a1.actor_id
JOIN actor a2 ON ap.actor2_id = a2.actor_id
ORDER BY ap.film_id, actor1_name, actor2_name;
```

Explanation:

CTE actor_pairs:

Performs a self-join on the film_actor table to find actor pairs in the same film.

The condition fa1.actor_id < fa2.actor_id avoids duplicate and self-pairs.

The final SELECT:

Joins the actor table twice to retrieve full names of both actors in the pair.

Returns a list of film IDs with actor name pairs who appeared together.

# 12. CTE for Recursive Search:
a. Implement a recursive CTE to find all employees in the staff table who report to a specific manager,
considering the reports_to column

```python
staff (
    staff_id INT PRIMARY KEY,
    first_name VARCHAR(50),
    last_name VARCHAR(50),
    reports_to INT  -- references staff_id of the manager
)
```

Ans:>


```python
-- Replace `1` with the staff_id of the manager you want to search under
WITH RECURSIVE staff_hierarchy AS (
    -- Anchor member: the top-level manager
    SELECT
        staff_id,
        first_name,
        last_name,
        reports_to
    FROM staff
    WHERE staff_id = 1  -- specific manager's ID

    UNION ALL

    -- Recursive member: find employees reporting to the current level
    SELECT
        s.staff_id,
        s.first_name,
        s.last_name,
        s.reports_to
    FROM staff s
    INNER JOIN staff_hierarchy sh ON s.reports_to = sh.staff_id
)

-- Final result
SELECT *
FROM staff_hierarchy;
```

Explanation:

The CTE staff_hierarchy starts from a specific manager (e.g., staff_id = 1).

The recursive part joins staff on reports_to, pulling in all employees who report up the chain.

This structure works even for multi-level hierarchies, showing all direct and indirect reports.
